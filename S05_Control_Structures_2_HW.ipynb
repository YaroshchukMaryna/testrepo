<div style="color:inherit;line-height:1.5;">
<p><code>room_capacities</code> contains room capacities in the order checked; <code>0</code> means that a room is unavailable. Write a function <code>first_suitable_room(room_capacities, required_capacity)</code> that returns the first available capacity that is at least <code>required_capacity</code>, or <code>-1</code> if none is found.</p>
<p>Example: <code>first_suitable_room([0, 20, 50, 80], 45)</code> returns <code>50</code>.</p>
<ul>
<li>Skip unavailable rooms with <code>continue</code>.</li>
<li>Stop at the first suitable room with <code>break</code>. A room whose capacity equals <code>required_capacity</code> is suitable.</li>
<li>Use loop <code>else</code> if no room matches.</li>
</ul>
</div>
<p style="color:inherit;"><strong>Points:</strong> 9</p>
</div>


```python


def first_suitable_room(room_capacities, required_capacity):
    for capacity in room_capacities:
        if capacity == 0:
            continue

        if capacity >= required_capacity:
            break
    else:
        return -1

    return capacity
    # TODO
```

<p>A lab technician enters <code>number_of_readings</code> temperature readings. Write a function <code>average_valid_temperatures(number_of_readings)</code> that asks the user for readings from 15 through 30 and returns their average.</p>
<p>Example: for <code>number_of_readings = 2</code>, the inputs <code>35</code>, <code>20</code>, <code>22</code> return <code>21.0</code>; <code>35</code> is invalid and must be entered again.</p>
<p>Assume <code>number_of_readings</code> is a positive integer and the user enters numeric values. The inclusive valid interval is 15 to 30.</p>
<ul>
<li>Read each temperature with <code>input()</code> and convert it with <code>float()</code>.</li>
<li>Use an outer <code>for</code> loop for exactly <code>number_of_readings</code> valid readings.</li>
<li>Use an inner <code>while</code> loop to ask again as many times as needed while the current reading is outside 15 through 30.</li>
<li>Add valid readings to a total and return the numeric average. Rounding the result to two or more decimal places is acceptable.</li>
</ul>
</div>
<p style="color:inherit;"><strong>Points:</strong> 10</p>
</div>


```python
def average_valid_temperatures(number_of_readings):
    total = 0.0
    for i in range(number_of_readings):
        temperature = float(input("Enter temperature: "))

        while temperature < 15 or temperature > 30:
            temperature = float(input("Invalid! Enter again: "))

        total += temperature
    return total / number_of_readings    

    # TODO
```
